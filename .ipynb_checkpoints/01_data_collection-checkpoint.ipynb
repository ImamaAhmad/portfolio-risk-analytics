```python
import pandas as pd
import numpy as np
import yfinance as yf
import matplotlib.pyplot as plt
import seaborn as sns
print("Everything works!")
```

```python
tickers = ["SPY", "VXUS", "EEM", "TLT", "GLD", "VNQ"]

prices = yf.download(
    tickers,
    start="2015-01-01",
    auto_adjust=True,
    progress=False
)

prices.head()
```

```python
weights_equal = pd.Series(
    1 / len(returns.columns),
    index=returns.columns
)

weights_equal
```

```python
portfolio_returns = returns.dot(weights_equal)

portfolio_returns.head()
```

```python
portfolio_annual_return = portfolio_returns.mean() * 252
portfolio_annual_volatility = portfolio_returns.std() * np.sqrt(252)
portfolio_sharpe = portfolio_annual_return / portfolio_annual_volatility

print(f"Annual Return: {portfolio_annual_return:.2%}")
print(f"Annual Volatility: {portfolio_annual_volatility:.2%}")
print(f"Sharpe Ratio: {portfolio_sharpe:.2f}")
```

```python
from scipy.optimize import minimize
```

```python
mean_returns = returns.mean() * 252
cov_matrix = returns.cov() * 252

n_assets = len(returns.columns)

def portfolio_stats(weights):
    portfolio_return = np.dot(weights, mean_returns)
    portfolio_volatility = np.sqrt(
        np.dot(weights.T, np.dot(cov_matrix, weights))
    )
    sharpe = portfolio_return / portfolio_volatility
    
    return portfolio_return, portfolio_volatility, sharpe
```

```python
constraints = {
    "type": "eq",
    "fun": lambda weights: np.sum(weights) - 1
}

bounds = tuple((0, 1) for _ in range(n_assets))

initial_weights = np.array([1 / n_assets] * n_assets)

result = minimize(
    lambda weights: -portfolio_stats(weights)[2],
    initial_weights,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints
)

optimal_weights = pd.Series(
    result.x,
    index=returns.columns
)

optimal_weights
```

```python
optimal_return, optimal_volatility, optimal_sharpe = portfolio_stats(
    optimal_weights.values
)

print(f"Optimized Annual Return: {optimal_return:.2%}")
print(f"Optimized Annual Volatility: {optimal_volatility:.2%}")
print(f"Optimized Sharpe Ratio: {optimal_sharpe:.2f}")
```

```python
comparison = pd.DataFrame({
    "Equal Weight": [
        portfolio_annual_return,
        portfolio_annual_volatility,
        portfolio_sharpe
    ],
    "Optimized": [
        optimal_return,
        optimal_volatility,
        optimal_sharpe
    ]
}, index=["Annual Return", "Annual Volatility", "Sharpe Ratio"])

comparison
```

```python
comparison.to_csv("data/processed/portfolio_comparison.csv")
```

```python
cumulative_returns = (1 + portfolio_returns).cumprod()

running_max = cumulative_returns.cummax()

drawdown = (cumulative_returns - running_max) / running_max

max_drawdown = drawdown.min()

print(f"Maximum Drawdown: {max_drawdown:.2%}")
```

```python
plt.figure(figsize=(12, 5))

plt.plot(drawdown, color="red")

plt.title("Equal-Weight Portfolio Drawdown")
plt.ylabel("Drawdown")
plt.xlabel("Date")

plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()
```

```python
var_95 = returns.dot(weights_equal).quantile(0.05)

print(f"95% Daily Historical VaR: {var_95:.2%}")
```

```python
portfolio_daily_returns = portfolio_returns.dropna()

var_95 = portfolio_daily_returns.quantile(0.05)

cvar_95 = portfolio_daily_returns[
    portfolio_daily_returns <= var_95
].mean()

print(f"95% VaR: {var_95:.2%}")
print(f"95% CVaR: {cvar_95:.2%}")
```

```python
cumulative_returns = (1 + portfolio_daily_returns).cumprod()

running_max = cumulative_returns.cummax()

drawdown = (cumulative_returns / running_max) - 1

print(f"Maximum Drawdown: {drawdown.min():.2%}")
```

```python
risk_metrics = pd.Series({
    "Annual Return": portfolio_annual_return,
    "Annual Volatility": portfolio_annual_volatility,
    "Sharpe Ratio": portfolio_sharpe,
    "95% VaR": var_95,
    "95% CVaR": cvar_95,
    "Maximum Drawdown": drawdown.min()
})

risk_metrics.to_csv("data/processed/risk_metrics.csv")

risk_metrics
```

```python
split_date = "2022-01-01"

train_returns = returns[returns.index < split_date]
test_returns = returns[returns.index >= split_date]

print("Training period:", train_returns.index.min(), "to", train_returns.index.max())
print("Testing period:", test_returns.index.min(), "to", test_returns.index.max())
```

```python
train_mean = train_returns.mean() * 252
train_cov = train_returns.cov() * 252

def train_portfolio_stats(weights):
    ret = np.dot(weights, train_mean)
    vol = np.sqrt(weights.T @ train_cov @ weights)
    sharpe = ret / vol
    return ret, vol, sharpe

train_result = minimize(
    lambda w: -train_portfolio_stats(w)[2],
    initial_weights,
    method="SLSQP",
    bounds=bounds,
    constraints=constraints
)

train_weights = pd.Series(
    train_result.x,
    index=train_returns.columns
)

train_weights
```

```python
test_portfolio_returns = test_returns.dot(train_weights)

test_return = test_portfolio_returns.mean() * 252
test_volatility = test_portfolio_returns.std() * np.sqrt(252)
test_sharpe = test_return / test_volatility

print(f"Out-of-Sample Return: {test_return:.2%}")
print(f"Out-of-Sample Volatility: {test_volatility:.2%}")
print(f"Out-of-Sample Sharpe: {test_sharpe:.2f}")
```

```python
oos_results = pd.Series({
    "Out-of-Sample Return": test_return,
    "Out-of-Sample Volatility": test_volatility,
    "Out-of-Sample Sharpe": test_sharpe
})

oos_results.to_csv("data/processed/out_of_sample_results.csv")

oos_results
```

```python
comparison_extended = pd.DataFrame({
    "Equal Weight": [
        portfolio_annual_return,
        portfolio_annual_volatility,
        portfolio_sharpe
    ],
    "Optimized In-Sample": [
        optimal_return,
        optimal_volatility,
        optimal_sharpe
    ],
    "Optimized Out-of-Sample": [
        test_return,
        test_volatility,
        test_sharpe
    ]
}, index=[
    "Annual Return",
    "Annual Volatility",
    "Sharpe Ratio"
])

comparison_extended
```
